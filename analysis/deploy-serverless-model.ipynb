```python
from sagemaker.huggingface import HuggingFaceModel
import sagemaker
from sagemaker.serverless import ServerlessInferenceConfig
```

```python
role = sagemaker.get_execution_role()
print(role)
```

```python
hub = {
    'HF_MODEL_ID': 'BIaoo/lca-qwen3-embedding',  # 你的 HuggingFace 模型 ID
    'HF_TASK': 'feature-extraction'  # 任务类型，如 text-classification, question-answering 等
}
```

```python
huggingface_model = HuggingFaceModel(
    env=hub,
    role=role,
    transformers_version='4.51.3',
    pytorch_version='2.6.0',
    py_version='py312',
)
```

```python

serverless_config = ServerlessInferenceConfig(
    memory_size_in_mb=4096,
    max_concurrency=1
)
```

```python
predictor = huggingface_model.deploy(
    serverless_inference_config=serverless_config
)
```

```python
from sagemaker.huggingface import HuggingFacePredictor
import sagemaker

# Endpoint 名称
endpoint_name = "huggingface-pytorch-inference-2026-01-12-09-56-55-646"

# 创建 predictor 对象连接到现有 endpoint
predictor = HuggingFacePredictor(
    endpoint_name=endpoint_name,
    sagemaker_session=sagemaker.Session()
)
```

```python
# 发送请求
data = {
    "inputs": ["This is a test sentence for embedding generation."]
}

response = predictor.predict(data)
print(response)
```
